# Healthcare Access

## Load Data

```python
def data_loader(data_type, table_name):
    path = f'data/{data_type}/{table_name}-eng/{table_name}.csv'
    df = pd.read_csv(path)
    return df
```

```python
healthcare_df = data_loader('health-indicators', '13100096')
healthcare_df.head()
```

- `value` is in *'thousands'* of people if `unit` is 'Number' and in '%' if `unit` is 'Percent'.

## Filter Status

```python
print(healthcare_df.shape[0])
```

```python
# filter for NaN or 'E' values in status
healthcare_df = healthcare_df[healthcare_df['status'].isin(['E', np.nan])]
```

## Filter Recent Data 

```python
# change ref_date to integer
healthcare_df['ref_date'] = pd.to_numeric(healthcare_df['ref_date'], errors='coerce')
```

```python
# filter ref_date >= 2018
healthcare_df = healthcare_df[healthcare_df['ref_date'] >= 2018]
```

```python
print(f"{healthcare_df.shape[0]} rows after filtering by ref_date >= 2018")
```

## Drop Missing Values 

```python
# Check NA value counts and proportions
na_counts = healthcare_df.isna().sum()
na_proportions = healthcare_df.isna().mean()
na_summary = pd.DataFrame({'count': na_counts, 'proportion': na_proportions})
na_summary
```

## Feature Engineering

```python
hc_renamed_df['indicators'].unique()
```

```python
fe_df = hc_renamed_df.copy()
```

```python
indicator_to_col = {
    # Perceived health
    'Perceived health, very good or excellent': 'Perceived_Health',
    'Perceived health, fair or poor': 'Perceived_Health',
    # Perceived mental health
    'Perceived mental health, very good or excellent': 'Perceived_Mental_Health',
    'Perceived mental health, fair or poor': 'Perceived_Mental_Health',
    # Life stress
    'Perceived life stress, most days quite a bit or extremely stressful': 'Life_Stress',
    # BMI
    'Body mass index, adjusted self-reported, adult (18 years and over), overweight': 'BMI_Adult',
    'Body mass index, adjusted self-reported, adult (18 years and over), obese': 'BMI_Adult',
    'Body mass index, self-reported, youth (12 to 17 years old), overweight or obese': 'BMI_Youth',
    # Chronic conditions
    'Arthritis (15 years and over)': 'Chronic_Condition',
    'Diabetes': 'Chronic_Condition',
    'Asthma': 'Chronic_Condition',
    'Chronic obstructive pulmonary disease (COPD; 35 years and over)': 'Chronic_Condition',
    'High blood pressure': 'Chronic_Condition',
    'Mood disorder': 'Chronic_Condition',
    'Anxiety disorder': 'Chronic_Condition',
    # Smoking
    'Current smoker, daily or occasional': 'Smoking_Status',
    'Current smoker, daily': 'Smoking_Status',
    # Alcohol/drug use
    'Heavy drinking': 'Alcohol_Drug_Use',
    'Cannabis use, past 12 months': 'Alcohol_Drug_Use',
    'Cannabis frequency of use in the past 12 months, daily or almost daily': 'Alcohol_Drug_Use',
    # Physical activity
    'Self-reported physical activity, 150 minutes per week, adult (18 years and over)': 'Physical_Activity',
    'Self-reported physical activity, average 60 minutes per day, youth (12 to 17 years old)': 'Physical_Activity',
    # Breastfeeding
    'Breast milk feeding initiation': 'Breastfeeding',
    'Exclusive breastfeeding, at least 6 months': 'Breastfeeding',
    # Belonging
    'Sense of belonging to local community, somewhat strong or very strong': 'Belonging',
    # Life satisfaction
    'Life satisfaction, satisfied or very satisfied': 'Life_Satisfaction',
    # Healthcare provider
    'Has a regular healthcare provider': 'Healthcare_Provider',
    # Immunization
    'Influenza immunization in the past 12 months': 'Immunization',
    # Nutrition
    'Fruit and vegetable consumption, 5 times or more per day': 'Nutrition',
    # Vaping
    'Ever used e-cigarette or vaping device': 'Vaping',
    'Used e-cigarette or vaping device, past 30 days': 'Vaping'
}

# create empty columns for each category
for col in set(indicator_to_col.values()):
    fe_df[col] = np.nan

# populate the relevant column per row
for i, row in fe_df.iterrows():
    indicator = row['indicators']
    col = indicator_to_col.get(indicator)
    value = indicator.split(',',1)[1] if ',' in indicator else indicator

    if col:
        fe_df.at[i, col] = value
```

```python
fe_df.shape[0], hc_renamed_df.shape[0]
```

```python
# drop original indicators column
fe_df = fe_df.drop(columns=['indicators'])
```
