# Healthcare Access

## Load Data

```python
def data_loader(data_type, table_name):
    path = f'data/{data_type}/{table_name}-eng/{table_name}.csv'
    df = pd.read_csv(path)
    return df
```

```python
healthcare_df = data_loader('health-indicators', '13100096')
healthcare_df.head()
```

```python
# Check data types
healthcare_df.info()
```

```python
print(f"{healthcare_df.shape[0]} rows")
```

- `value` represents percentage/number of people with 'unmet' healthcare needs.
- `value` is in *'thousands'* of people if `unit` is 'Number' and in '%' if `unit` is 'Percent'.

## Rename Column Names


```python
healthcare_df.columns
```

```python
# change column names to lower case
healthcare_df.columns = healthcare_df.columns.str.lower()
```

```python
# replace spaces with underscores
healthcare_df.columns = healthcare_df.columns.str.replace(' ', '_')
```

```python
# rename 'sex' to 'gender' for consistency
healthcare_df = healthcare_df.rename(columns={'sex':'gender'})
```

```python
healthcare_df.columns
```

## Statistics of the Data

```python
# statistics of the numerical columns
healthcare_df.describe()
```

Since standard deviation is 0 and min and max are the same, it means that all values in the column are the same.

## Drop unnecessary numerical columns


```python
healthcare_df = healthcare_df.drop(columns=['uom_id', 'scalar_id', 'symbol', 'decimals'])
```

```python
# categorical columns
healthcare_df.describe(include=['object'])
```

```python
# unique values in the categorical columns
for col in ['geo','age_group', 'gender', 'indicators','characteristics','uom']:
    print(f'{col}: {healthcare_df[col].nunique()} unique values')
    print(healthcare_df[col].unique())
    print()
```

## Drop unnecessary categorical columns

```python
# drop unnecessary categorical columns
for col in ['dguid', 'uom',	'scalar_factor','vector','coordinate','status','terminated']:
    if col in healthcare_df.columns:
        healthcare_df = healthcare_df.drop(columns=[col])
```

```python
healthcare_df.head()
```

## Filter Recent Data 

```python
# change ref_date to integer
healthcare_df['ref_date'] = pd.to_numeric(healthcare_df['ref_date'], errors='coerce')
```

```python
# filter ref_date >= 2018
healthcare_df = healthcare_df[healthcare_df['ref_date'] >= 2018]
```

```python
print(f"{healthcare_df.shape[0]} rows after filtering by ref_date >= 2018")
```

## Drop Missing Values 

```python
# Check NA value counts and proportions
na_counts = healthcare_df.isna().sum()
na_proportions = healthcare_df.isna().mean()
na_summary = pd.DataFrame({'count': na_counts, 'proportion': na_proportions})
na_summary
```

```python
# Drop rows with NA values
hc_cleaned_df = healthcare_df.dropna()
print(f"Dropped {healthcare_df.shape[0] - hc_cleaned_df.shape[0]} row(s) with NA values")
```

```python
print(hc_cleaned_df.shape[0], "rows remaining")
```

## Feature Engineering

```python
fe_df = hc_cleaned_df.copy()
```

```python
# split 'geo' column into 'location' and 'age'
def split_geo(geo_value):
    if ',' in geo_value:
        loc, age = geo_value.split(',', 1)
        return pd.Series([loc.strip(), age.strip()])
    else:
        return pd.Series([geo_value.strip(), None])

fe_df[['location', 'age']] = fe_df['geo'].apply(split_geo)
```

```python
fe_df.shape[0], hc_cleaned_df.shape[0]
```

## Replace 'nan' to np.nan

```python
fe_df = fe_df.replace('nan', np.nan)
```
