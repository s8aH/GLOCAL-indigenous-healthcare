# Health Indicators

## Load Data

```python
def data_loader(data_type, table_name):
    path = f'../../data/{data_type}/{table_name}-eng/{table_name}.csv'
    df = pd.read_csv(path)
    return df
```

```python
hi_df = data_loader('health-indicators', '13100924')
hi_df.head()
```

```python
# Check data types
hi_df.info()
```

The `value` column shows the percentage of individuals in each Indigenous group (and gender category) who selected a specific healthcare experience

## Rename Column Names


```python
hi_df.columns
```

```python
# change column names to lower case
hi_df.columns = hi_df.columns.str.lower()
```

```python
# replace spaces with underscores
hi_df.columns = hi_df.columns.str.replace(' ', '_')
```

```python
hi_df.columns
```

## Statistics of the Data

```python
# statistics of the numerical columns
hi_df.describe()
```

Since standard deviation is 0 and min and max are the same, it means that all values in the column are the same.

Since count is 0, it means there are no non-null values in the column.

Some values are > 100 which is not possible for percentages. This indicates that there are some errors in the data.

```python
# only keep values <= 100
hi_filtered_df = hi_df[hi_df['value'] <= 100]
```

```python
# categorical columns
hi_filtered_df.describe(include=['object'])
```

Since unique is 1, it means that all values in these columns are the same. So they are not useful for analysis.

```python
# unique values in the categorical columns
for col in ['ref_date', 'geo', 'age_group', 'sex', 'indigenous_identity','indicators', 'characteristics']:
    print(f'{col}: {hi_filtered_df[col].nunique()} unique values')
    print(hi_filtered_df[col].unique())
    print()
```

## Drop Unnecessary Columns

```python
selected_cols = ['ref_date', 'geo', 'age_group', 'sex', 'indigenous_identity',
       'indicators', 'characteristics', 'value']
```

```python
hi_filtered_df = hi_filtered_df[selected_cols]
```

```python
hi_filtered_df.columns
```

## Keep Only Percentage Characteristics

```python
# filter characteristics == 'Percent'
hi_filtered_df1 = hi_filtered_df[hi_filtered_df['characteristics'] == 'Percent']
print(f"{hi_filtered_df.shape[0]} - {hi_filtered_df1.shape[0]} = {hi_filtered_df.shape[0] - hi_filtered_df1.shape[0]}")
```

## Drop Missing Values 

```python
# Check NA value counts and proportions
na_counts = hi_filtered_df1.isna().sum()
na_proportions = hi_filtered_df1.isna().mean()
na_summary = pd.DataFrame({'count': na_counts, 'proportion': na_proportions})
na_summary
```

# Dataframe Pivoting

```python
df = hi_filtered_df1.copy()
```

```python
df['indicators'].unique()
```

```python
# mapping
subgroup_map = {
    'Perceived health': 'Perceived Health Status',
    'Perceived mental health': 'Perceived Mental Health Status',
    'Perceived life stress': 'Life Stress',
    'Body mass index': 'BMI Category',
    'Current smoker': 'Smoking Status',
    'Heavy drinking': 'Alcohol Use',
    'Never had alcoholic drinks in the past 12 months': 'Alcohol Use',
    'At least one underlying chronic health condition': 'Chronic Health Condition Count',
    'Two or more chronic health conditions': 'Chronic Health Condition Count',
    'Arthritis': 'Chronic/Physical Health Condition',
    'Diabetes': 'Chronic/Physical Health Condition',
    'High blood pressure': 'Chronic/Physical Health Condition',
    'Heart disease': 'Chronic/Physical Health Condition',
    'Effects of a stroke': 'Chronic/Physical Health Condition',
    'Anxiety disorder': 'Chronic/Physical Health Condition',
    'Mood disorder': 'Chronic/Physical Health Condition',
    'Has ever had cancer': 'Chronic/Physical Health Condition',
    'Sense of belonging to local community': 'Social Wellbeing',
    'Life satisfaction': 'Social Wellbeing',
    'Has a regular healthcare provider': 'Healthcare Access',
    'Household food insecure': 'Healthcare Access'
}

# create subgroup columns and fill with None
for col in set(subgroup_map.values()):
    df[col] = None

# populate each new column as described
def fill_subgroup(row):
    for key, col in subgroup_map.items():
        if row['indicators'].startswith(key):
            row[col] = row['indicators']
    return row

df_wide = df.apply(fill_subgroup, axis=1)
```

# Rename Columns

```python
# rename sex column as gender for consistency
df_wide = df_wide.rename(columns={'sex':'gender'})
```

```python
# replace space with underscore in column names
df_wide.columns = df_wide.columns.str.replace(' ', '_')
```

```python
# lowercase all column names
df_wide.columns = df_wide.columns.str.lower()
```

```python
df = df_wide.copy()
```
