## Feature Engineering

```python
hc_cleaned_df['access_to_and_use_of_health_care_services'].unique()
```

```python
# group mapping
group_map = {
    'Total, regular medical doctor status': 'regular_medical_doctor_status',
    'Total, contact with health professional in the last 12 months': 'contact_with_health_professional_12mo',
    'Total, received dental care in the last three years': 'received_dental_care_3yr',
    'Total, health care required in the last 12 months but not received': 'healthcare_required_not_received_12mo'
}

# Assign group to each row
def assign_group_col(df, col='access_to_and_use_of_health_care_services', group_map=group_map):
    group_col = []
    current_group = None
    for val in df[col]:
        if val in group_map:
            current_group = group_map[val]
        group_col.append(current_group)
    df['__temp_group'] = group_col
    return df

# For each group, create a new column, fill with the original value if in that group, else NaN
def split_by_group(df, col='access_to_and_use_of_health_care_services'):
    for group in df['__temp_group'].unique():
        if group is None:
            continue
        df[group] = np.where(df['__temp_group'] == group, df[col], np.nan)
    return df

hc_feature_eng_df = assign_group_col(hc_cleaned_df)
hc_feature_eng_df = split_by_group(hc_feature_eng_df)

# drop the temporary group column
hc_feature_eng_df = hc_feature_eng_df.drop(columns='__temp_group')
```

```python
# drop access_to_and_use_of_health_care_services column
hc_feature_eng_df = hc_feature_eng_df.drop(columns='access_to_and_use_of_health_care_services')
```
